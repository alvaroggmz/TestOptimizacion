```python

#https://www.youtube.com/watch?v=lq8FvxjEX0o
#https://www.youtube.com/watch?v=oG5GFOjkJ3k 

import pygame
import math

# Configuración inicial
ANCHO, ALTO = 600, 400
pygame.init()
pantalla = pygame.display.set_mode((ANCHO, ALTO))
reloj = pygame.time.Clock()

# Definir colores
BLANCO = (255, 255, 255)
NEGRO = (0, 0, 0)
ROJO = (200, 50, 50)

# Longitud de los segmentos del brazo
L1, L2, L3 = 100, 75, 50  # Tres segmentos

def calcular_puntos(base, ang1, ang2, ang3):
    """Calcula las posiciones de cada articulación"""
    x1 = base[0] + L1 * math.cos(math.radians(ang1))
    y1 = base[1] + L1 * math.sin(math.radians(ang1))
    
    x2 = x1 + L2 * math.cos(math.radians(ang1 + ang2))
    y2 = y1 + L2 * math.sin(math.radians(ang1 + ang2))
    
    x3 = x2 + L3 * math.cos(math.radians(ang1 + ang2 + ang3))
    y3 = y2 + L3 * math.sin(math.radians(ang1 + ang2 + ang3))

    return [(int(x1), int(y1)), (int(x2), int(y2)), (int(x3), int(y3))]

# Posición de la base del brazo
base = (150, 300)
ang1, ang2, ang3 = 45, -30, 20  # Ángulos iniciales

ejecutando = True
while ejecutando:
    pantalla.fill(BLANCO)
    
    for evento in pygame.event.get():
        if evento.type == pygame.QUIT:
            ejecutando = False
    
    # Obtener posición del mouse y calcular ángulos
    mx, my = pygame.mouse.get_pos()
    ang1 = math.degrees(math.atan2(my - base[1], mx - base[0]))  # Ángulo base
    ang2 = -30  # Fijo para simplicidad
    ang3 = 20   # Fijo para simplicidad

    # Calcular posiciones
    puntos = calcular_puntos(base, ang1, ang2, ang3)

    # Dibujar brazo
    pygame.draw.line(pantalla, NEGRO, base, puntos[0], 5)
    pygame.draw.line(pantalla, NEGRO, puntos[0], puntos[1], 5)
    pygame.draw.line(pantalla, NEGRO, puntos[1], puntos[2], 5)
    pygame.draw.circle(pantalla, ROJO, puntos[2], 8)

    pygame.display.flip()
    reloj.tick(30)

pygame.quit()

```
